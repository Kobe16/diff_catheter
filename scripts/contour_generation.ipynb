```python
# Generate heart shape using parametric equations
t = np.linspace(0, 2 * np.pi, 1000)
x = 16 * np.sin(t) ** 3
y = 13 * np.cos(t) - 5 * np.cos(2 * t) - 2 * np.cos(3 * t) - np.cos(4 * t)

# Plot heart shape
plt.figure(figsize=(6.4, 4.8),dpi=300)
plt.fill(x, y, '#7E57C2')  # Fill the heart with color
plt.axis("equal")
plt.axis("off")  # Hide axis
# plt.title("Heart Shape", fontsize=14)

# Show plot
plt.show()

```

```python
def draw_rotated_rectangle(center, width, height, angle, color='#64B5F6'):
    """
    Draw a filled rotated rectangle with no outline on a white background.

    :param center: (x, y) - Center of the rectangle
    :param width: Width of the rectangle
    :param height: Height of the rectangle
    :param angle: Rotation angle in degrees (counterclockwise)
    :param color: Fill color of the rectangle
    """
    # Define rectangle corners relative to the center
    corners = np.array([
        [-width / 2, -height / 2],
        [ width / 2, -height / 2],
        [ width / 2,  height / 2],
        [-width / 2,  height / 2]
    ])
    
    # Rotation matrix
    theta = np.radians(angle)
    rotation_matrix = np.array([
        [np.cos(theta), -np.sin(theta)],
        [np.sin(theta),  np.cos(theta)]
    ])
    
    # Apply rotation and translation
    rotated_corners = np.dot(corners, rotation_matrix.T) + center

    # Plot rotated rectangle
    plt.figure(figsize=(6.4,4.8), facecolor='white', dpi=300)
    plt.fill(rotated_corners[:, 0], rotated_corners[:, 1], color=color, alpha=1.0)
    plt.axis("equal")
    plt.axis("off")  # Hide axes

    # Show plot
    plt.show()

# Example usage: Draw a rotated rectangle
draw_rotated_rectangle(center=(0, 0), width=4, height=2, angle=30, color='#1565C0')
```
